# IATA Dangerous Goods Regulations (DGR) — 65th Edition
## Section 4.4: Special Provisions — Review

**Source:** IATA DGR 65th Edition (2024), digitalpublications.iata.org  
**Reference URL:** `DGR-6065-65` / Section `DGR-4.4`  
**Effective date:** 1 January 2024

---

## 1. Context: Structure of Section 4

Section 4 of the IATA DGR — *Identification* — is divided into four subsections:

| Subsection | Title |
|---|---|
| 4.1 | Proper Shipping Name |
| 4.2 | List of Dangerous Goods |
| 4.3 | Numerical Cross-Reference List |
| **4.4** | **Special Provisions** |

Section 4.2 contains the core Dangerous Goods List (DGL). Column M of that list carries alpha-numeric Special Provision codes. When Column M references a code, the shipper must consult Section 4.4 for additional requirements that apply *on top of* (or instead of) the standard entry conditions.

---

## 2. Purpose of Section 4.4

Special Provisions (SPs) provide subsidiary rules for specific substances or articles. They may:

- **Restrict** transport (e.g., require additional approval documents).
- **Relax** standard requirements (e.g., allow higher net quantity per package).
- **Clarify** classification or labelling for an entry.
- **Permit** reclassification as "not restricted" cargo under defined conditions.

Where an SP aligns with an UN Model Regulations special provision, the UN number appears in parentheses after the IATA SP code.

> **Key rule (8.1.6.9.4):** An SP code must only be noted in the *Authorizations* column of the Shipper's Declaration when the shipper has actually *applied* that provision to the shipment — not merely because the code appears in the DGL against that entry.

---

## 3. Special Provisions Requiring Declaration (Authorizations Column)

The following 16 SPs must be shown in the Authorizations column of the Shipper's Declaration for Dangerous Goods (per 8.1.6.9.4) **as applicable**:

```
A1  A2  A4  A5  A51  A81  A88  A99
A130  A190  A191  A201  A202  A211  A212  A331
```

Later editions (65th +) added: **A176, A224, A225, A334**.

## 4. Key Special Provisions — Descriptions

The table below summarises commonly referenced SPs. Full regulatory text is in the official 65th Edition publication.

| SP Code | Applies to | Description |
|---|---|---|
| **A1** | Goods forbidden PAX, permitted CAO | Transport on passenger aircraft permitted if the shipper holds competent authority approval(s). Copy of approval must be attached to the Shipper's Declaration. Must be noted in the Authorizations column. |
| **A2** | Goods requiring operator approval | Transport permitted subject to operator approval(s). Copy of approval must accompany the Shipper's Declaration. Must be noted in the Authorizations column. |
| **A4** | Various entries | Specifies additional quantity or packaging conditions for the referenced entry. |
| **A5** | Various entries | Specifies conditions under which the entry may be transported with relaxed requirements. |
| **A51** | UN 2794 / UN 2795 (Aircraft batteries) | Packages containing aircraft batteries may exceed the standard 30 kg net quantity limit for passenger aircraft, up to 100 kg net quantity per package. |
| **A81** | Various | Articles or substances assigned to this entry must meet additional design or test criteria. |
| **A88** | Prototype lithium batteries/cells | Applies only to prototype or small-production-run batteries not yet tested to applicable UN standards. Must be noted in Authorizations column when applied. |
| **A99** | Various | Denotes additional documentation or marking requirements for the specific entry. |
| **A130** | Various | Provides for specific handling or stowage conditions beyond standard requirements. |
| **A176** | Metal hydride storage systems | SP number must be noted in the Authorizations column of the Shipper's Declaration when this provision is applied (amended in 64th edition). |
| **A190** | Various | Additional packing or quantity restrictions apply. |
| **A191** | Various | Specifies subsidiary hazard label requirements or exemptions. |
| **A201** | Various | Permits transport under conditions that deviate from the standard entry. |
| **A202** | Various | Applies additional net quantity or package limits. |
| **A211** | Various | Specifies marking or documentation beyond standard requirements. |
| **A212** | Various | Provides additional packing group or classification conditions. |
| **A221** | UN 1002 (Compressed Air) | Nitrogen-oxygen mixtures at specified concentrations may be transported as UN 1002 without requiring a Division 5.1 subsidiary hazard label. *(Added 64th edition)* |
| **A223** | UN 2990 / UN 3072 (Life-saving appliances) | Appliances packed in rigid outer packagings with only Division 2.2 gas cartridges for device activation may be shipped as "not restricted" cargo. *(Added 64th edition)* |
| **A224** | UN 3548 (Articles, misc. dangerous goods) | Articles containing environmentally hazardous substances may ship under Packing Instruction 975, exempting SP A2. Must be noted in Authorizations column. *(Added 64th edition)* |
| **A225** | UN 3538 (Articles, non-flammable non-toxic gas) | Articles with Division 2.2 gas (no subsidiary hazard) may ship under Packing Instruction 222, exempting SP A2. Must be noted in Authorizations column. *(Added 64th edition)* |
| **A331** | Various | Specifies conditions for transport as limited quantity or provides quantity cap relaxation. |
| **A803** | UN 1486 (Potassium Nitrate, PG III) | Substances assigned to this entry must be packaged in PG II UN-specification packaging despite the PG III packing group designation. |

## 5. Changes Introduced in the 65th Edition (2024) — Section 4.4

The 65th edition incorporated all IATA Dangerous Goods Board amendments effective 1 January 2024. Key Section 4.4 changes from the 64th→65th edition transition include:

### 5.1 Amended Provisions

| SP | Change |
|---|---|
| **A1** | Clarified that when dangerous goods are offered under an approval, "A1" must be noted in the Authorizations column of the Shipper's Declaration. |
| **A2** | Same clarification as A1 — explicit requirement to note in the Authorizations column. |
| **A176** | Modified wording: metal hydride storage system shipments must explicitly note the SP number in the Authorizations column. |

### 5.2 New Provisions Added

| SP | Entry | Description |
|---|---|---|
| **A221** | UN 1002 | Permits nitrogen-oxygen mixtures at specified concentrations without Division 5.1 subsidiary hazard label. |
| **A223** | UN 2990, UN 3072 | Life-saving appliances with only Division 2.2 gas cartridges may ship as "not restricted." |
| **A224** | UN 3548 | Articles with environmentally hazardous substances may use PI 975, exempting SP A2. |
| **A225** | UN 3538 | Articles with Division 2.2 gas (no subsidiary hazard) may use PI 222, exempting SP A2. |

## 7. Practical Compliance Checklist

When preparing a dangerous goods shipment where Column M of the DGL shows a Section 4.4 SP code:

1. **Locate** the SP text in Section 4.4.
2. **Determine** whether the SP applies to your specific shipment scenario.
3. **If applied**, note the SP code in the *Authorizations* column of the Shipper's Declaration (required for: A1, A2, A4, A5, A51, A81, A88, A99, A130, A176, A190, A191, A201, A202, A211, A212, A224, A225, A331).
4. **Attach** any required documents (e.g., competent authority approval for A1/A2).
5. **Do not** list an SP code on the Declaration unless the provision has actually been applied — erroneous declarations can cause shipment delays or rejection.

````python
# Quick reference: SPs that MUST appear in the Shipper's Declaration Authorizations column

DECLARATION_REQUIRED_SPS = [
    "A1",   # Goods forbidden PAX / permitted CAO under approval
    "A2",   # Goods requiring operator approval
    "A4",
    "A5",
    "A51",  # Aircraft batteries exceeding 30 kg limit
    "A81",
    "A88",  # Prototype lithium batteries
    "A99",
    "A130",
    "A176", # Metal hydride storage systems (added 64th ed.)
    "A190",
    "A191",
    "A201",
    "A202",
    "A211",
    "A212",
    "A224", # Articles containing environmentally hazardous substances (added 64th ed.)
    "A225", # Articles containing non-flammable non-toxic gas (added 64th ed.)
    "A331",
]

print("Special Provisions requiring notation in the Authorizations column:")
print(", ".join(DECLARATION_REQUIRED_SPS))
print(f"\nTotal: {len(DECLARATION_REQUIRED_SPS)} provisions")
````

````python
# Compliance helper: check if a given SP code requires declaration notation

def requires_declaration_notation(sp_code: str) -> bool:
    """Return True if the SP must be noted in the Authorizations column."""
    return sp_code.upper() in DECLARATION_REQUIRED_SPS


# Examples
for code in ["A1", "A51", "A88", "A221", "A223", "A803"]:
    flag = requires_declaration_notation(code)
    print(f"  SP {code}: {'MUST be noted in Authorizations column' if flag else 'No declaration notation required'}")
````
